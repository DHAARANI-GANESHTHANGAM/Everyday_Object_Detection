```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
results = model.predict(source="https://ultralytics.com/images/bus.jpg", save=True, conf=0.25)
print("Saved to:", results[0].save_dir)

r = results[0]
ms = r.speed["preprocess"] + r.speed["inference"] + r.speed["postprocess"]
print(f"{ms:.1f} ms per image -> ~{1000/ms:.0f} FPS")
```
